```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

    def age_in_5_years(self):
        return self.age + 5

her = Person("Alice", 30)
him =Person("Bob", 25)
him.greet()  # Output: Hello, my name is Alice and I am 30
him.age_in_5_years()
```
