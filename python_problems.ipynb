```python
#FACTORIAL
def fac(n):
    if n==0:
        return 1
    else:
        return n*fac(n-1)
fac(7)
```

```python
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True
is_prime(11)
is_prime(565787)
```

```python
for i in range(1,100):
    if is_prime(i):
        if str(i)!=str(i)[::-1]:
            print(i, end = " ")
```

```python
#Fibonacci series
def fib(n):
    if n<=1:
        return n
    else:
        return fib(n-1)+fib(n-2)
fib(7)
```
